```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
### fetching the data online for the multiple linear regression
from sklearn.datasets import fetch_california_housing
```

```python
california = fetch_california_housing()
```

```python
california.keys()
```

```python
california
```

```python
print(california.DESCR)
```

```python
california.target_names ## This is our output features
```

```python
### Prepare the dataframe
df = pd.DataFrame(california.data, columns=california.feature_names)
```

```python
df.head()  ## all are the in dependent features
```

```python
df['Price'] = california.target ## dependent features or target
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(), annot=True)
plt.Figure(figsize=(20,10))
```

```python
X = df.drop(df[['Price']], axis=1)
y = df['Price']

'''
Or 
X = df.iloc[:,:-1] take all except last one
y = df.iloc[:,-1] take only last one

'''
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=10)
```

```python
X_train.shape, y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

## Model Training

```python
from sklearn.linear_model import LinearRegression
linreg = LinearRegression()
```

```python
linreg.fit(X_train, y_train)
```

```python
linreg.coef_   ## 8 coefficients 
```

```python
linreg.intercept_
```

```python
## Prediction
y_pred = linreg.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print(mean_absolute_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```

```python
residuals = y_test-y_pred
```

```python
plt.scatter(y_pred, residuals)
```

```python
plt.scatter(y_test, y_pred)
plt.xlabel("Test data")
plt.ylabel('Predicted data')
```

```python
sns.displot(residuals, kind='kde')
```
